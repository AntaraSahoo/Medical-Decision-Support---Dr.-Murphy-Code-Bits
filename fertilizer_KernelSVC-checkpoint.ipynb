```python
import pandas as pd
import numpy as np
```

```python
dataset = pd.read_csv("C:\\Users\\KIIT\\Desktop\\agriculture project\\Fertilizer Prediction.csv")
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values
```

```python
from sklearn.preprocessing import LabelEncoder
le1 = LabelEncoder()
X[:,3] = le1.fit_transform(X[:,3])

le2 = LabelEncoder()
X[:,4] = le2.fit_transform(X[:,4])
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 0)
print(type(X_train))
print(type(y_test))
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
from sklearn.svm import SVC
classifier = SVC(kernel = 'sigmoid', random_state = 0)
classifier.fit(X_train, y_train)
```

```python
#c = classifier.predict(sc.transform([[26,52,38,'Sandy','Maize',37,0,0]]))
```

```python
y_pred = classifier.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(y_test, y_pred)
# print(cm)
am = accuracy_score(y_test, y_pred)
print (am)
```
