```python
#import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# read an excel file and convert into a dataframe object 
df = pd.DataFrame(pd.read_excel("C:\\Users\\KIIT\\Desktop\\Thickness_datasets\\ann_steel_data.xlsx")) 
# show the dataframe 
df
```

```python
#extracting input and output parameters
X = df.iloc[2: , 1:5].values
y = df.iloc[2: , 5].values
print(X)
```

```python
#normalising values
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X = sc.fit_transform(X)
```

```python
# Training the Random Forest Regression model on the whole dataset
from sklearn.ensemble import RandomForestRegressor
regressor = RandomForestRegressor(n_estimators = 50, random_state = 2)
regressor.fit(X, y)
```

```python
#predicted values
y_pred = regressor.predict(X)
print(y_pred.reshape(-1,1))
```

```python
# plotting a graph of predicted data v/s given data
plt.plot(range(0,164), y_pred, color = 'blue')
plt.plot(range(0,164), y, color = 'red')
plt.title('Given Data v/s Predicted Data')
plt.xlabel('No.of observations')
plt.ylabel('Thickness(nm)')
plt.legend(['Predicted Data', 'Given Data'])
plt.show()
```

```python
# finding r2 score - if it is near to 1, the model is perfect
from sklearn.metrics import r2_score
r2 = r2_score(y, y_pred) 
print('r2 score for this is', r2)
```
