```python
import tensorflow
import matplotlib.pyplot as plt 
import numpy as np
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D
from tensorflow.keras.layers import MaxPooling2D
from tensorflow.keras.layers import Flatten
from tensorflow.keras.layers import Dense
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
train_datagen = ImageDataGenerator(rescale = 1./255,
                                 zoom_range = 0.2,
                                 vertical_flip = True,
                                 horizontal_flip = True)

training_set = train_datagen.flow_from_directory("C://Users//KIIT//Desktop//minor project//datasets//Lung cancer CT scan//train",
                                                target_size = (64,64),
                                                class_mode = 'sparse')

test_datagen = ImageDataGenerator(rescale = 1./255)

test_set = test_datagen.flow_from_directory("C://Users//KIIT//Desktop//minor project//datasets//Lung cancer CT scan//test",
                                                target_size = (64,64),
                                                            class_mode = 'sparse')
```

```python

#test_image = "C://Users//KIIT//Desktop//0b8dabb7-5f1b-4fdc-b3fa-30b289707b90___JR_FrgE.S 3047_90deg.JPG"
#test_image = np.array((test_image))
#plt.imshow(test_image[0])
```

```python
classifier = Sequential()

#first dense layer
classifier.add(Conv2D(filters = 32, kernel_size = (3, 3), input_shape = [64, 64, 3], activation = 'relu'))
classifier.add(MaxPooling2D(pool_size = (2, 2), strides = 2)) 

#second dense layer
classifier.add(Conv2D(filters = 32, kernel_size = (3, 3), input_shape = [64, 64, 3], activation = 'relu'))
classifier.add(MaxPooling2D(pool_size = (2, 2), strides = 2)) 
classifier.add(Flatten())
```

```python
classifier.add(Dense(units = 128, activation = 'relu'))
classifier.add(Dense(units = 32, activation = 'relu'))
classifier.add(Dense(units = 16, activation = 'relu'))
classifier.add(Dense(units = 4, activation = 'softmax')) #put entire data into probablity
classifier.compile(optimizer = 'adam', loss = 'sparse_categorical_crossentropy', metrics = ['sparse_categorical_accuracy']) # compiling = stochastic gradient 
```

```python
classifier.fit_generator(training_set,
                         epochs = 25, 
                         validation_data = test_set)
```
