```python
import tensorflow as tf
from keras.preprocessing.image import ImageDataGenerator
```

```python
tf.__version__
```

```python
train_datagen = ImageDataGenerator(rescale = 1./255,
                                   shear_range = 0.2,
                                   zoom_range = 0.2,
                                   horizontal_flip = True)
training_set = train_datagen.flow_from_directory('Corn_dataset/training_set',
                                                 target_size = (64, 64),
                                                 batch_size = 32,
                                                 class_mode = 'sparse')
```

```python
testing_datagen = ImageDataGenerator(rescale = 1./255)
testing_set = testing_datagen.flow_from_directory('Corn_Dataset/testing_set',
                                            target_size = (64, 64),
                                            batch_size = 32,
                                            class_mode = 'sparse')
```

```python
cnn = tf.keras.models.Sequential()
```

```python
cnn.add(tf.keras.layers.Conv2D(filters=32, kernel_size=3, activation='relu', input_shape=[64, 64, 3]))
```

```python
cnn.add(tf.keras.layers.MaxPool2D(pool_size=2, strides=2))
```

```python
cnn.add(tf.keras.layers.Conv2D(filters=32, kernel_size=3, activation='relu'))
cnn.add(tf.keras.layers.MaxPool2D(pool_size=2, strides=2))
```

```python
cnn.add(tf.keras.layers.Flatten())
```

```python
cnn.add(tf.keras.layers.Dense(units=128, activation='relu'))
```

```python
cnn.add(tf.keras.layers.Dense(units=4, activation='sigmoid'))
```

```python
cnn.compile(optimizer = 'adam', loss = 'sparse_categorical_crossentropy', metrics = ['sparse_categorical_accuracy'])
```

```python
cnn.fit(x = training_set, validation_data = testing_set, epochs = 25)
```

```python
import numpy as np
from keras.preprocessing import image
test_image = image.load_img('Corn_Dataset/test/CornCommonRust3.jpg', target_size = (64, 64))
test_image = image.img_to_array(test_image)
test_image = test_image/255
test_image = np.expand_dims(test_image, axis = 0)
result = cnn.predict_classes(test_image)
training_set.class_indices
if result == 0:
    prediction= 'Cercospora leaf spot Gray leaf spo'
elif result == 1:
    prediction = 'common rust'
elif result == 2:
    prediction = 'healthy'
elif result == 3:
    prediction = 'northern leaf blight'
```
